### Métodos y atributos útiles de pd.DataFrame

```python
# Importa dependencias
import pandas as pd
```

```python
# Creamos un DataFrame de ejemplo
cursos = pd.DataFrame(
     {"alumnos": [12, 7, 9], "precio": [49.0, 120.0, 35.0]},
     index=["SQL", "Pandas", "Excel"],
 )
print(cursos)
```

```python
# La traspuesta de cursos
print(cursos.T)
```

```python
# Localizar un valor
print("Valor en ubicación exacta: Fila -> SQL, Columna -> alumnos")
print(cursos.loc["SQL", "alumnos"])

print("\nValor en rango de ubicación: Fila -> SQL hasta Pandas, Columna -> alumnos")
print(cursos.loc["SQL":"Pandas", "alumnos"])

print("\nValor en ubicación exacta: Fila -> SQL, Columna -> alumnos")
print(cursos.at["SQL", "alumnos"])

# at no funciona para slicing (Error)
#print(cursos.at["SQL":"Pandas", "alumnos"])
```

```python
 # Solo existe el flag allows_duplicate_labels y es True por defecto
 cursos.flags
```

```python
print(cursos.iloc[0])
```

```python
 # Para filtrar por condición en fila y seleccionar columna

 # Forma 1: 
 print(cursos[cursos["alumnos"] > 8]["precio"])

 # Forma 2: 
 print(cursos.loc[cursos["alumnos"] > 8, "precio"])
```
