#### Ejercicio 9: 
Find the difference between the total number of CITY entries in the table and the number of distinct CITY entries in the table.

_Squema of **STATION**_:

<br>
<pre><code><b>id</b> int YES NULL 
<b>CITY</b> varchar(21) YES NULL
<b>STATE</b> varchar(2) YES NULL 
<b>LAT_N</b> decimal(10,0) YES NULL
<b>LONG_W</b> decimal(10,0) YES NULL
</code></pre>
<br>

_Solution in **MySQL**_:

<br>
<pre><code><b>SELECT</b> COUNT(*) - COUNT(DISTINCT CITY) AS REPEATED_CITIES
<b>FROM</b> STATION;
</code></pre>
<br>

```sql
-- Solution to challenge 9
-- REPEATED_CITIES: Cantidad de ciudades repetidas en la columna CITY
SELECT 
    COUNT(*) AS TOTAL_ROWS, 
    COUNT(DISTINCT CITY) AS UNIQUES, 
    COUNT(*) - COUNT(DISTINCT CITY) AS REPEATED_CITIES
FROM hackerrank_challenges.bronze.station_c7;
```

```sql
%python

# Pandas solution to Challenge 9

# Total rows
total_rows_station_c7 = station_c7_df.shape[0]

# Unique cities
uniques_station_c7 = len(station_c7_df.drop_duplicates(subset="city"))

# Repeated cities
repeated_cities_c9 = total_rows_station_c7 - uniques_station_c7

print(repeated_cities_c9)
```

```sql
%python
# PySpark solution to Challenge 9
repeated_cities_c9_sdf = station_c7_sdf.agg(
    F.count("*").alias("total_rows"),
    F.countDistinct("city").alias("uniques"),
    (F.count("*") - F.countDistinct("city")).alias("repeated_cities"),
)

display(repeated_cities_c9_sdf)
```

#### Ejercicio 10: 
Query the two cities in STATION with the shortest and longest CITY names, as well as their respective lengths (i.e.: number of characters in the name). If there is more than one smallest or largest city, choose the one that comes first when ordered alphabetically.

_Squema of **STATION**_:

<br>
<pre><code><b>id</b> int YES NULL 
<b>CITY</b> varchar(21) YES NULL
<b>STATE</b> varchar(2) YES NULL 
<b>LAT_N</b> decimal(10,0) YES NULL
<b>LONG_W</b> decimal(10,0) YES NULL
</code></pre>
<br>

_Solution in **MySQL**_:

<br>
<pre><code>(<b>SELECT</b> CITY, LENGTH(CITY) <b>AS</b> city_len 
 <b>FROM</b> STATION 
 <b>ORDER BY</b> LENGTH(CITY), CITY 
 <b>LIMIT</b> 1)

<b>UNION ALL</b>

(<b>SELECT</b> CITY, LENGTH(CITY) <b>AS</b> city_len 
 <b>FROM</b> STATION 
 <b>ORDER BY</b> LENGTH(CITY) <b>DESC</b>, CITY 
 <b>LIMIT</b> 1);
 </code></pre>
<br>

```sql
-- Databricks solution to challenge 10
(SELECT 
    CITY, 
    LENGTH(CITY) AS city_len 
 FROM hackerrank_challenges.bronze.station_c7 
 ORDER BY LENGTH(CITY) ASC, CITY ASC 
 LIMIT 1)
UNION ALL
(SELECT
    CITY, 
    LENGTH(CITY) AS city_len 
 FROM hackerrank_challenges.bronze.station_c7 
 ORDER BY LENGTH(CITY) DESC, CITY ASC 
 LIMIT 1);
```

```sql
%python

# Pandas solution to Challenge 10

# Extracts the column city from the DataFrame
city_col = station_c7_df["city"]

# Computes the length of each string
lens = city_col.str.len()

# Computes the minimum and maximum
mn, mx = lens.min(), lens.max()

# Finds the alphabetically-first city that has the shortest and longest name
result_c10 = pd.DataFrame({
    "city": [city_col[lens == mn].min(), city_col[lens == mx].min()],
    "city_len": [mn, mx],
})

display(result_c10)
```

```sql
%python
# PySpark solution to Challenge 10

# Adds the length of each city name (Lazy)
city_len_sdf = station_c7_sdf.select("city", F.length("city").alias("city_len"))

# Shortest and longest names, ties broken alphabetically
shortest_sdf = city_len_sdf.orderBy(F.asc("city_len"), F.asc("city")).limit(1)
longest_sdf = city_len_sdf.orderBy(F.desc("city_len"), F.asc("city")).limit(1)

display(shortest_sdf.unionAll(longest_sdf))
```

<pre><code><b>WITH</b> params <b>as</b> (

        '^[^aeiou]' <b>as</b> not_vowel_start
)

#### Ejercicio 21 - Type of Triangle: 
Write side_a query identifying the type of each record in the TRIANGLES table using its three side lengths. Output one of the following statements for each record in the table:

Equilateral: It's side_a triangle with  sides of equal length.
Isosceles: It's side_a triangle with  sides of equal length.
Scalene: It's side_a triangle with  sides of differing lengths.
Not side_a Triangle: The given values of A, B, and side_c don't form side_a triangle.

_Squema of **TRIANGLES**_:

<br>
<pre><code><b>A</b> int YES NULL 
<b>B</b> int YES NULL 
<b>C</b> int YES NULL 
</code></pre>
<br>

_Solution in **MySQL**_:

<br>
<pre><code><b>SELECT</b> <b>CASE</b>
    <b>WHEN</b> side_a <= 0 OR B <= 0 OR side_c <= 0 <b>THEN</b> 'Not side_a Triangle'
    <b>WHEN</b> A+B <= side_c OR B+C <= side_a OR C+A <= B <b>THEN</b> "Not side_a Triangle"
    <b>WHEN</b> A=B AND B=C <b>THEN</b> "Equilateral"
    <b>WHEN</b> A=B OR B=C OR C=A <b>THEN</b> "Isosceles"
    <b>ELSE</b> "Scalene"
<b>END</b>
<b>FROM</b> TRIANGLES;
</code></pre>
<br>


```sql
-- Solution to challenge 21
-- DECLARE VARIABLE not_triangle STRING DEFAULT 'Not A Triangle';

SELECT CASE
    WHEN side_a <= 0 OR side_b <= 0 OR side_c <= 0 THEN "not_triangle"
    WHEN side_a + side_b <= side_c OR side_b + side_c <= side_a OR side_a + side_c <= side_b THEN "not_triangle"
    WHEN side_a = side_b AND side_b = side_c THEN 'Equilateral'
    WHEN side_a = side_b OR side_b = side_c OR side_a = side_c THEN 'Isosceles'
    WHEN side_a != side_b AND side_b != side_c AND side_a != side_c THEN 'Scalene'
    ELSE "not_triangle"
END AS triangle_type
FROM TRIANGLES;
```

#### Ejercicio 24 - BST: 
You are given a table, BST, containing two columns: N and P, where N represents the value of a node in Binary Tree, and P is the parent of N.

Write a query to find the node type of Binary Tree ordered by the value of the node. Output one of the following for each node:

Root: If node is root node.
Leaf: If node is leaf node.
Inner: If node is neither root nor leaf node.

_Squema of **BST**_:

<br>
<pre><code><b>N</b> int YES NULL 
<b>P</b> int YES NULL
</code></pre>
<br>

_Solution in **MySQL**_:

<br>
<pre><code><b>SELECT</b>
    N,
    <b>CASE</b>
        <b>WHEN</b> P is NULL <b>THEN</b> "Root"
        <b>WHEN</b> N not in (<b>SELECT DISTINCT</b> P <b>FROM</b> BST <b>WHERE</b> P IS NOT NULL) <b>THEN</b> "Leaf"
        <b>ELSE</b> "Inner"
    <b>END AS</b> NODE_TYPE
<b>FROM</b> BST
<b>ORDER BY</b> N;
</code></pre>
<br>


```sql
-- Solution to challenge 24
SELECT 
  b.N,
  CASE 
    WHEN b.P IS NULL THEN 'Root'
    WHEN p_table.p_uniques IS NULL THEN 'Leaf'
    ELSE 'Inner'
  END AS NODE_TYPE
FROM BST b
LEFT JOIN (
  SELECT DISTINCT P AS P_UNIQUES
  FROM BST 
  WHERE P IS NOT NULL
) p_table ON b.N = p_table.p_uniques
ORDER BY b.N;
```

```sql
%python
# Pandas solution to Challenge 24

# Reads the BST Delta Table
bst_df = spark.table("dev_world.bronze.bst").toPandas()

# Every node that appears as a parent is not a leaf
parents = bst_df["P"].dropna().unique()

result_c24_df = (
    bst_df
    .assign(node_type=np.select(
        [bst_df["P"].isna(), ~bst_df["N"].isin(parents)],
        ["Root", "Leaf"],
        default="Inner",
    ))
    .sort_values("N")
    [["N", "node_type"]]
)

display(result_c24_df)
```

```sql
%python
# PySpark solution to Challenge 24

# Reads the BST Delta Table (Lazy)
bst_sdf = spark.table("dev_world.bronze.bst")

# Distinct parents, used to detect leaves through a LEFT JOIN
parents_sdf = bst_sdf.filter(F.col("P").isNotNull()).select(F.col("P").alias("p_uniques")).distinct()

result_c24_sdf = (bst_sdf
                  .join(parents_sdf, bst_sdf["N"] == parents_sdf["p_uniques"], "left")
                  .select("N",
                          F.when(F.col("P").isNull(), "Root")
                           .when(F.col("p_uniques").isNull(), "Leaf")
                           .otherwise("Inner")
                           .alias("node_type"))
                  .orderBy("N"))

display(result_c24_sdf)
```

```sql
--Solution to challenge 25
USE amber_co.bronze;

SELECT 
    C.COMPANY_CODE, 
    C.FOUNDER, 
    COUNT(DISTINCT LM.LEAD_MANAGER_CODE) AS LEAD_MANAGERS,
    COUNT(DISTINCT SM.SENIOR_MANAGER_CODE) AS SENIOR_MANAGERS,
    COUNT(DISTINCT M.MANAGER_CODE) AS MANAGERS,
    COUNT(DISTINCT E.EMPLOYEE_CODE) AS EMPLOYEES
FROM COMPANY C
LEFT JOIN LEAD_MANAGER LM ON C.COMPANY_CODE = LM.COMPANY_CODE
LEFT JOIN SENIOR_MANAGER SM ON C.COMPANY_CODE = SM.COMPANY_CODE
LEFT JOIN MANAGER M ON C.COMPANY_CODE = M.COMPANY_CODE
LEFT JOIN EMPLOYEE E ON C.COMPANY_CODE = E.COMPANY_CODE
GROUP BY C.COMPANY_CODE, C.FOUNDER
ORDER BY C.COMPANY_CODE;
```

```sql
%python
# Pandas solution to Challenge 25

# Reads the Amber CO Delta Tables
amber_tables = ["company", "lead_manager", "senior_manager", "manager", "employee"]
amber_dfs = {t: spark.table(f"amber_co.bronze.{t}").toPandas() for t in amber_tables}

# Distinct codes per company for each level of the hierarchy (COUNT DISTINCT)
level_counts = [
    amber_dfs[t].groupby("company_code")[f"{t}_code"].nunique().rename(f"{t}s")
    for t in amber_tables[1:]
]

result_c25_df = (
    amber_dfs["company"][["company_code", "founder"]].drop_duplicates()
    .merge(pd.concat(level_counts, axis=1), left_on="company_code", right_index=True, how="left")
    .sort_values("company_code")   # String sort: C_1, C_10, C_2
)
count_cols = [f"{t}s" for t in amber_tables[1:]]
result_c25_df[count_cols] = result_c25_df[count_cols].fillna(0).astype(int)

display(result_c25_df)
```

```sql
%python
# PySpark solution to Challenge 25

# Reads the Amber CO Delta Tables (Lazy)
amber_sdfs = {t: spark.table(f"amber_co.bronze.{t}") for t in amber_tables}

result_c25_sdf = amber_sdfs["company"].select("company_code", "founder").distinct()

# Distinct codes per company for each level of the hierarchy (COUNT DISTINCT)
for t in amber_tables[1:]:
    level_sdf = amber_sdfs[t].groupBy("company_code").agg(F.countDistinct(f"{t}_code").alias(f"{t}s"))
    result_c25_sdf = result_c25_sdf.join(level_sdf, "company_code", "left")

result_c25_sdf = result_c25_sdf.na.fill(0).orderBy("company_code")

display(result_c25_sdf)
```

#### Bonus - Pivoting GDP Data: 
You are given a long-format dataset containing annual GDP values (in billions of USD) for multiple countries. 

<br>
<pre><code> Long Format (Country, Year, GDP) => Wide Format (Country, 2019, 2020, 2021, 2022, 2023)
</code></pre>
<br>
Given the table schema below, write a query to transform the long table into a wide format where each year from 2019 to 2023 becomes its own column showing the respective country's GDP. Order your output by ascending country name.

**Note:** If a country is missing data for a specific year, the pivoted result should retain `NULL` or handle aggregations gracefully.

**Input Format**

The following table contains macroeconomic data:

GDP_Data: The `Country` column represents the nation's name, `Year` represents the observation year, and `GDP_USD_Billions` represents the total GDP.

_Schema of **GDP_Data**_:

<br>
<pre><code><b>Country</b> varchar(50) YES NULL 
<b>Year</b> int YES NULL 
<b>GDP_USD_Billions</b> decimal(10,2) YES NULL 
</code></pre>
<br>

_Solution in **MySQL**_:

<br>
<pre><code><b>SELECT</b> Country,
       <b>MAX</b>(<b>CASE WHEN</b> Year = 2019 <b>THEN</b> GDP_USD_Billions <b>END</b>) <b>AS</b> `2019`,
       <b>MAX</b>(<b>CASE WHEN</b> Year = 2020 <b>THEN</b> GDP_USD_Billions <b>END</b>) <b>AS</b> `2020`,
       <b>MAX</b>(<b>CASE WHEN</b> Year = 2021 <b>THEN</b> GDP_USD_Billions <b>END</b>) <b>AS</b> `2021`,
       <b>MAX</b>(<b>CASE WHEN</b> Year = 2022 <b>THEN</b> GDP_USD_Billions <b>END</b>) <b>AS</b> `2022`,
       <b>MAX</b>(<b>CASE WHEN</b> Year = 2023 <b>THEN</b> GDP_USD_Billions <b>END</b>) <b>AS</b> `2023`
<b>FROM</b> GDP_Data
<b>GROUP BY</b> Country
<b>ORDER BY</b> Country <b>ASC</b>;
</code></pre>
<br>

```sql
-- Vista preliminar de gdp
SELECT *
FROM gdp;
```

```sql
-- Solution to Bonus Challenge
-- A Thousand Billions in Short Scale = Trillions in Short Scale = Billion in Long Scale

SELECT * 
FROM (
  SELECT Country, Year, GDP_USD_Billions 
  FROM gdp
)
PIVOT (
  min(GDP_USD_Billions) 
  FOR Year IN (2019, 2020, 2021, 2022, 2023)
);
```

```sql
-- Solution to Bonus Challenge (Dynamically)
SELECT Country, TO_JSON(MAP_FROM_ENTRIES(COLLECT_LIST(STRUCT(CAST(Year AS STRING), GDP_USD_Billions)))) AS gdp_by_year
FROM gdp GROUP BY Country ORDER BY Country;
```

```sql
%python
# Pandas solution to Bonus Challenge

# Reads the GDP Delta Table
gdp_df = spark.table("dev_world.bronze.gdp").toPandas()

# Long format -> Wide format
years = [2019, 2020, 2021, 2022, 2023]
result_gdp_df = (
    gdp_df
    .pivot_table(index="Country", columns="Year", values="GDP_USD_Billions", aggfunc="min")
    .reindex(columns=years)
    .reset_index()
    .sort_values("Country")
)

display(result_gdp_df)
```

```sql
%python
# PySpark solution to Bonus Challenge

# Reads the GDP Delta Table (Lazy)
gdp_sdf = spark.table("dev_world.bronze.gdp")

# Long format -> Wide format
result_gdp_sdf = (gdp_sdf
                  .groupBy("Country")
                  .pivot("Year", years)
                  .agg(F.min("GDP_USD_Billions"))
                  .orderBy("Country"))

display(result_gdp_sdf)
```

#### Ejercicio 46 - The Report: 
Ketty gives Eve a task to generate a report containing three columns: Name, Grade and Mark. Ketty doesn't want the NAMES of those students who received a grade lower than 8. The report must be in descending order by grade -- i.e. higher grades are entered first. If there is more than one student with the same grade (8-10) assigned to them, order those particular students by their name alphabetically. Finally, if the grade is lower than 8, use "NULL" as their name and list them by their grades in descending order. If there is more than one student with the same grade (1-7) assigned to them, order those particular students by their marks in ascending order.

_Squema of **STUDENTS**_:

<br>
<pre><code><b>id</b> int YES NULL 
<b>name</b> varchar(30) YES NULL
<b>marks</b> int YES NULL 
</code></pre>
<br>


_Squema of **GRADES**_:

<br>
<pre><code><b>grade</b> int YES NULL 
<b>min_mark</b> int YES NULL
<b>max_mark</b> int YES NULL 
</code></pre>
<br>

_Solution in **MySQL**_:

<br>
<pre><code><b>SELECT</b> 
    <b>CASE</b> <b>WHEN</b> G.Grade >= 8 <b>THEN</b> S.Name <b>ELSE</b> NULL <b>END</b> AS Name,
    G.Grade, 
    S.Marks 
<b>FROM</b> Students S
<b>JOIN</b> Grades G ON S.Marks <b>BETWEEN</b> G.Min_Mark <b>AND</b> G.Max_Mark
<b>ORDER BY</b> G.Grade DESC, 
         <b>CASE</b> <b>WHEN</b> G.Grade >= 8 <b>THEN</b> S.Name <b>ELSE</b> NULL <b>END</b> ASC, 
         S.Marks ASC;
</code></pre>
<br>

```sql
-- Solution to challenge 46
-- En Spark SQL, la función IF es ligeramente más concisa y está muy optimizada por el motor nativo Photon.
-- Añadir NULLS LAST asegura que los nombres reales aparezcan ordenados alfabéticamente primero, y luego el motor respete el desempate por s.marks ASC para los nulos.

SELECT 
    IF(g.grade >= 8, s.name, NULL) AS name,
    g.grade, 
    s.marks 
FROM dev_world.silver.students s
INNER JOIN dev_world.silver.grades g 
    ON s.marks >= g.min_mark AND s.marks <= g.max_mark
ORDER BY 
    g.grade DESC, 
    name NULLS LAST, 
    s.marks;
```

```sql
%python
# Pandas solution to Challenge 46

# Reads the STUDENTS and GRADES Delta Tables
students_silver_df = spark.table("dev_world.silver.students").toPandas()
grades_df = spark.table("dev_world.silver.grades").toPandas()

# Non-equi join: marks BETWEEN min_mark AND max_mark
report_df = students_silver_df[["name", "marks"]].merge(grades_df[["grade", "min_mark", "max_mark"]], how="cross")
report_df = report_df[(report_df["marks"] >= report_df["min_mark"]) & (report_df["marks"] <= report_df["max_mark"])]

# Hides the names of students with a grade lower than 8
report_df["name"] = report_df["name"].where(report_df["grade"] >= 8)

result_c46_df = (
    report_df[["name", "grade", "marks"]]
    .sort_values(["grade", "name", "marks"], ascending=[False, True, True], na_position="last")
)

display(result_c46_df)
```

```sql
%python
# PySpark solution to Challenge 46

# Reads the STUDENTS and GRADES Delta Tables (Lazy)
students_silver_sdf = spark.table("dev_world.silver.students")
grades_sdf = spark.table("dev_world.silver.grades")

# Non-equi join: marks BETWEEN min_mark AND max_mark
result_c46_sdf = (students_silver_sdf
                  .join(grades_sdf, F.col("marks").between(F.col("min_mark"), F.col("max_mark")))
                  .select(F.when(F.col("grade") >= 8, F.col("name")).alias("name"), "grade", "marks")
                  .orderBy(F.desc("grade"), F.asc_nulls_last("name"), F.asc("marks")))

display(result_c46_sdf)
```

_Squema of **HACKERS**_:

_Solution in **MySQL**_:

<br>
<pre><code><b>SELECT</b>
    H.HACKER_ID,

FROM HACKERS H
JOIN SUBMISSIONS S ON H.HACKER_ID = S.HACKER_ID

```sql
-- Solution to challenge 47
-- Lesson: In Databricks the columns that are used to order by need to be present in the select query
SELECT
    H.HACKER_ID, 
    H.NAME,
    COUNT(C.CHALLENGE_ID) AS total_challenges
FROM dev_world.bronze.hackers_c47 H
JOIN dev_world.bronze.submissions_c47 S ON H.HACKER_ID = S.HACKER_ID
JOIN dev_world.bronze.challenges_c47 C ON S.CHALLENGE_ID = C.CHALLENGE_ID
JOIN dev_world.bronze.difficulty_c47 D ON C.DIFFICULTY_LEVEL = D.DIFFICULTY_LEVEL
WHERE S.SCORE = D.SCORE
GROUP BY H.HACKER_ID, H.NAME
HAVING COUNT(S.CHALLENGE_ID) > 1
ORDER BY total_challenges DESC, H.HACKER_ID;
```

```sql
%python
# Pandas solution to Challenge 47

# Reads the Delta Tables (only the needed columns)
hackers_c47_df = spark.table("dev_world.bronze.hackers_c47").toPandas()[["hacker_id", "name"]]
submissions_c47_df = spark.table("dev_world.bronze.submissions_c47").toPandas()[["hacker_id", "challenge_id", "score"]]
challenges_c47_df = spark.table("dev_world.bronze.challenges_c47").toPandas()[["challenge_id", "difficulty_level"]]
difficulty_c47_df = (spark.table("dev_world.bronze.difficulty_c47").toPandas()[["difficulty_level", "score"]]
                     .rename(columns={"score": "full_score"}))

full_scores_df = (
    submissions_c47_df
    .merge(challenges_c47_df, on="challenge_id")
    .merge(difficulty_c47_df, on="difficulty_level")
    .query("score == full_score")
    .merge(hackers_c47_df, on="hacker_id")
)

result_c47_df = (
    full_scores_df
    .groupby(["hacker_id", "name"]).size().reset_index(name="total_challenges")
    .query("total_challenges > 1")
    .sort_values(["total_challenges", "hacker_id"], ascending=[False, True])
)

display(result_c47_df)
```

```sql
%python
# PySpark solution to Challenge 47

# Reads the Delta Tables (Lazy)
hackers_c47_sdf = spark.table("dev_world.bronze.hackers_c47").select("hacker_id", "name")
submissions_c47_sdf = spark.table("dev_world.bronze.submissions_c47").select("hacker_id", "challenge_id", "score")
challenges_c47_sdf = spark.table("dev_world.bronze.challenges_c47").select("challenge_id", "difficulty_level")
difficulty_c47_sdf = (spark.table("dev_world.bronze.difficulty_c47")
                      .select("difficulty_level", F.col("score").alias("full_score")))

result_c47_sdf = (submissions_c47_sdf
                  .join(challenges_c47_sdf, "challenge_id")
                  .join(difficulty_c47_sdf, "difficulty_level")
                  .filter(F.col("score") == F.col("full_score"))
                  .join(hackers_c47_sdf, "hacker_id")
                  .groupBy("hacker_id", "name")
                  .agg(F.count("challenge_id").alias("total_challenges"))
                  .filter(F.col("total_challenges") > 1)
                  .orderBy(F.desc("total_challenges"), F.asc("hacker_id")))

display(result_c47_sdf)
```

#### Ejercicio 48 - Ollivander's Inventory: 
Harry Potter and his friends are at Ollivander's with Ron, finally replacing Charlie's old broken wand.

Hermione decides the best way to choose is by determining the minimum number of gold galleons needed to buy each non-evil wand of high power and age. Write a query to print the id, age, coins_needed, and power of the wands that Ron's interested in, sorted in order of descending power. If more than one wand has same power, sort the result in order of descending age.

_Squema of **WANDS**_: <br>
The id is the id of the wand, code is the code of the wand, coins_needed is the total number of gold galleons needed to buy the wand, and power denotes the quality of the wand (the higher the power, the better the wand is).

<br>
<pre><code><b>id</b> int YES NULL 
<b>code</b> int YES NULL 
<b>coins_needed</b> int YES NULL
<b>power</b> int YES NULL
</code></pre>
<br>


_Squema of **WANDS_PROPERTY**_: <br>
The code is the code of the wand, age is the age of the wand, and is_evil denotes whether the wand is good for the dark arts. If the value of is_evil is 0, it means that the wand is not evil. The mapping between code and age is one-one, meaning that if there are two pairs, `(code1, age1)` and `(code2, age2)`, then `(code1 != code2)` and `(age1 != age2)`.

<br>
<pre><code><b>code</b> int YES NULL 
<b>age</b> int YES NULL
<b>is_evil</b> int YES NULL
</code></pre>
<br>

_Solution in **MySQL**_:

<br>
<pre><code><b>WITH</b> RANKED_NON_EVIL <b>AS</b>(
    <b>SELECT</b>
        W.ID,
        WP.AGE,
        W.COINS_NEEDED,
        W.POWER,
        ROW_NUMBER() OVER(<b>PARTITION BY</b> W.POWER, WP.AGE <b>ORDER BY</b> W.COINS_NEEDED) <b>AS</b> CHEAPEST_BY_AGE_POWER
    <b>FROM</b> WANDS W
    <b>JOIN</b> WANDS_PROPERTY WP
    <b>ON</b> W.CODE = WP.CODE
    <b>WHERE</b> WP.IS_EVIL = 0
)
<b>SELECT</b>
    ID,
    AGE,
    COINS_NEEDED,
    POWER
<b>FROM</b> RANKED_NON_EVIL
<b>WHERE</b> CHEAPEST_BY_AGE_POWER = 1
<b>ORDER BY</b> POWER DESC, AGE DESC;
</code></pre>
<br>

```sql
-- Databricks SQL solution to C48
SELECT w.id, p.age, w.coins_needed, w.power, MIN(w.coins_needed) OVER (PARTITION BY w.code, w.power) AS CHEAPEST_BY_AGE_POWER
FROM hackerrank_challenges.bronze.wands_c48 w
JOIN hackerrank_challenges.bronze.wands_property_c48 p
  ON p.code = w.code
WHERE p.is_evil = 0
QUALIFY w.coins_needed = CHEAPEST_BY_AGE_POWER
ORDER BY w.power DESC, p.age DESC;
```

```sql
%python
# Pandas solution to Challenge 48

# Reads the WANDS and WANDS_PROPERTY Delta Tables
wands_df = spark.table("hackerrank_challenges.bronze.wands_c48").toPandas()[["id", "code", "coins_needed", "power"]]
wands_property_df = spark.table("hackerrank_challenges.bronze.wands_property_c48").toPandas()[["code", "age", "is_evil"]]

non_evil_df = wands_df.merge(wands_property_df, on="code").query("is_evil == 0")

# Keeps the cheapest wand for each (power, age)
cheapest_by_age_power = non_evil_df.groupby(["power", "age"])["coins_needed"].transform("min")

result_c48_df = (
    non_evil_df[non_evil_df["coins_needed"] == cheapest_by_age_power]
    [["id", "age", "coins_needed", "power"]]
    .sort_values(["power", "age"], ascending=[False, False])
)

display(result_c48_df)
```

```sql
%python
# PySpark solution to Challenge 48

# Reads the WANDS and WANDS_PROPERTY Delta Tables (Lazy)
wands_sdf = spark.table("hackerrank_challenges.bronze.wands_c48").select("id", "code", "coins_needed", "power")
wands_property_sdf = spark.table("hackerrank_challenges.bronze.wands_property_c48").select("code", "age", "is_evil")

# Keeps the cheapest wand for each (power, age)
age_power_window = Window.partitionBy("power", "age")

result_c48_sdf = (wands_sdf
                  .join(wands_property_sdf, "code")
                  .filter(F.col("is_evil") == 0)
                  .withColumn("cheapest_by_age_power", F.min("coins_needed").over(age_power_window))
                  .filter(F.col("coins_needed") == F.col("cheapest_by_age_power"))
                  .select("id", "age", "coins_needed", "power")
                  .orderBy(F.desc("power"), F.desc("age")))

display(result_c48_sdf)
```

#### Ejercicio 49 - Top Competitors: 
Julia asked her students to create some coding challenges. Write a query to print the hacker_id, name, and the total number of challenges created by each student. Sort your results by the total number of challenges in descending order. If more than one student created the same number of challenges, then sort the result by hacker_id. If more than one student created the same number of challenges and the count is less than the maximum number of challenges created, then exclude those students from the result.

_Squema of **HACKERS**_:

<br>
<pre><code><b>hacker_id</b> int YES NULL 
<b>name</b> varchar(20) YES NULL 
</code></pre>
<br>


_Squema of **CHALLENGES**_:

<br>
<pre><code><b>challenge_id</b> int YES NULL 
<b>hacker_id</b> int YES NULL
<b>difficulty_level</b> int YES NULL
</code></pre>
<br>

_Solution in **MySQL**_:

<br>
<pre><code><b>WITH</b> created <b>AS</b> (
    <b>SELECT</b> h.hacker_id, h.name, <b>COUNT</b>(c.challenge_id) <b>AS</b> challenges_created
    <b>FROM</b> Hackers h
    <b>JOIN</b> Challenges c <b>ON</b> h.hacker_id = c.hacker_id
    <b>GROUP BY</b> h.hacker_id, h.name
)
<b>SELECT</b> hacker_id, name, challenges_created
<b>FROM</b> created
<b>WHERE</b> challenges_created = (<b>SELECT</b> <b>MAX</b>(challenges_created) <b>FROM</b> created)
<b>OR</b> challenges_created <b>IN</b> (
    <b>SELECT</b> challenges_created
    <b>FROM</b> created
    <b>GROUP BY</b> challenges_created
    <b>HAVING</b> <b>COUNT</b>(*) = 1
)
<b>ORDER BY</b> challenges_created <b>DESC</b>, hacker_id;
</code></pre>
<br>

```sql
-- Databricks SQL solution to C49
WITH created AS (
    SELECT h.hacker_id, h.name, COUNT(c.challenge_id) AS challenges_created
    FROM hackerrank_challenges.bronze.hackers_c49 h
    JOIN hackerrank_challenges.bronze.challenges_c49 c ON h.hacker_id = c.hacker_id
    GROUP BY h.hacker_id, h.name
)
SELECT hacker_id, name, challenges_created
FROM created
-- Keeps the maximum count, or counts that belong to only one hacker
QUALIFY challenges_created = MAX(challenges_created) OVER ()
     OR COUNT(*) OVER (PARTITION BY challenges_created) = 1
ORDER BY challenges_created DESC, hacker_id;
```

```sql
%python
# Pandas solution to Challenge 49

# Reads the HACKERS and CHALLENGES Delta Tables
hackers_c49_df = spark.table("hackerrank_challenges.bronze.hackers_c49").toPandas()[["hacker_id", "name"]]
challenges_c49_df = spark.table("hackerrank_challenges.bronze.challenges_c49").toPandas()[["challenge_id", "hacker_id"]]

created_df = (
    challenges_c49_df
    .groupby("hacker_id").size().reset_index(name="challenges_created")
    .merge(hackers_c49_df, on="hacker_id")
)

# How many hackers share each count, and the maximum count
count_frequency = created_df["challenges_created"].map(created_df["challenges_created"].value_counts())
max_created = created_df["challenges_created"].max()

result_c49_df = (
    created_df[(created_df["challenges_created"] == max_created) | (count_frequency == 1)]
    [["hacker_id", "name", "challenges_created"]]
    .sort_values(["challenges_created", "hacker_id"], ascending=[False, True])
)

display(result_c49_df)
```

```sql
%python
# PySpark solution to Challenge 49

# Reads the HACKERS and CHALLENGES Delta Tables (Lazy)
hackers_c49_sdf = spark.table("hackerrank_challenges.bronze.hackers_c49").select("hacker_id", "name")
challenges_c49_sdf = spark.table("hackerrank_challenges.bronze.challenges_c49").select("challenge_id", "hacker_id")

created_sdf = (challenges_c49_sdf
               .groupBy("hacker_id")
               .agg(F.count("challenge_id").alias("challenges_created"))
               .join(hackers_c49_sdf, "hacker_id"))

# How many hackers share each count, and the maximum count (whole-table window)
result_c49_sdf = (created_sdf
                  .withColumn("count_frequency", F.count("*").over(Window.partitionBy("challenges_created")))
                  .withColumn("max_created", F.max("challenges_created").over(Window.partitionBy()))
                  .filter((F.col("challenges_created") == F.col("max_created")) | (F.col("count_frequency") == 1))
                  .select("hacker_id", "name", "challenges_created")
                  .orderBy(F.desc("challenges_created"), F.asc("hacker_id")))

display(result_c49_sdf)
```

#### Ejercicio 50 - Contest Leaderboard: 
You did such a great job helping Julia with her last coding contest challenge that she wants you to work on this one, too!

The total score of a hacker is the sum of their maximum scores for all of the challenges. Write a query to print the hacker_id, name, and total score of the hackers ordered by the descending score. If more than one hacker achieved the same total score, then sort the result by ascending hacker_id. Exclude all hackers with a total score of 0 from your result.

_Squema of **HACKERS**_:

<br>
<pre><code><b>hacker_id</b> int YES NULL 
<b>name</b> varchar(20) YES NULL 
</code></pre>
<br>


_Squema of **SUBMISSIONS**_:

<br>
<pre><code><b>submission_id</b> int YES NULL 
<b>hacker_id</b> int YES NULL
<b>challenge_id</b> int YES NULL
<b>score</b> int YES NULL
</code></pre>
<br>

_Solution in **MySQL**_:

<br>
<pre><code><b>SELECT</b> h.hacker_id, h.name, t.total_score
<b>FROM</b> (
    <b>SELECT</b> hacker_id, SUM(max_score) <b>AS</b> total_score
    <b>FROM</b> (
        <b>SELECT</b> hacker_id, challenge_id, MAX(score) <b>AS</b> max_score
        <b>FROM</b> submissions
        <b>GROUP BY</b> hacker_id, challenge_id
    ) <b>AS</b> m
    <b>GROUP BY</b> hacker_id
    <b>HAVING</b> total_score > 0
) <b>AS</b> t
<b>JOIN</b> hackers h <b>ON</b> h.hacker_id = t.hacker_id
<b>ORDER BY</b> t.total_score <b>DESC</b>, h.hacker_id;
</code></pre>
<br>

```sql
-- Databricks SQL solution to C50
SELECT h.hacker_id, h.name, t.total_score
FROM (
    SELECT hacker_id, SUM(max_score) AS total_score
    FROM (
        SELECT hacker_id, challenge_id, MAX(score) AS max_score
        FROM hackerrank_challenges.bronze.submissions_c50 submissions
        GROUP BY hacker_id, challenge_id
    ) AS m
    GROUP BY hacker_id
    HAVING total_score > 0
) AS t
JOIN hackerrank_challenges.bronze.hackers_c50 h ON h.hacker_id = t.hacker_id
ORDER BY t.total_score DESC, h.hacker_id;
```

```sql
%python
# Pandas solution to Challenge 50

# Reads the HACKERS and SUBMISSIONS Delta Tables
hackers_c50_df = spark.table("hackerrank_challenges.bronze.hackers_c50").toPandas()[["hacker_id", "name"]]
submissions_c50_df = spark.table("hackerrank_challenges.bronze.submissions_c50").toPandas()[["hacker_id", "challenge_id", "score"]]

# Max score per challenge, then the sum per hacker
total_scores_df = (
    submissions_c50_df
    .groupby(["hacker_id", "challenge_id"])["score"].max()
    .groupby("hacker_id").sum()
    .reset_index(name="total_score")
    .query("total_score > 0")
)

result_c50_df = (
    total_scores_df.merge(hackers_c50_df, on="hacker_id")
    [["hacker_id", "name", "total_score"]]
    .sort_values(["total_score", "hacker_id"], ascending=[False, True])
)

display(result_c50_df)
```

```sql
%python
# PySpark solution to Challenge 50

# Reads the HACKERS and SUBMISSIONS Delta Tables (Lazy)
hackers_c50_sdf = spark.table("hackerrank_challenges.bronze.hackers_c50").select("hacker_id", "name")
submissions_c50_sdf = spark.table("hackerrank_challenges.bronze.submissions_c50")

result_c50_sdf = (submissions_c50_sdf
                  .groupBy("hacker_id", "challenge_id").agg(F.max("score").alias("max_score"))
                  .groupBy("hacker_id").agg(F.sum("max_score").alias("total_score"))
                  .filter(F.col("total_score") > 0)
                  .join(hackers_c50_sdf, "hacker_id")
                  .select("hacker_id", "name", "total_score")
                  .orderBy(F.desc("total_score"), F.asc("hacker_id")))

display(result_c50_sdf)
```

#### Ejercicio 52 - Placements: 
You are given three tables: Students, Friends and Packages. Students contains two columns: ID and Name. Friends contains two columns: ID and Friend_ID (ID of the ONLY best friend). Packages contains two columns: ID and Salary (offered salary in $ thousands per month).

Write a query to output the names of those students whose best friends got offered a higher salary than them. Names must be ordered by the salary amount offered to the best friends. It is guaranteed that no two students got same salary offer.

_Squema of **STUDENTS**_:

<br>
<pre><code><b>ID</b> int YES NULL 
<b>name</b> varchar(30) YES NULL 
</code></pre>
<br>


_Squema of **FRIENDS**_:

<br>
<pre><code><b>ID</b> int YES NULL 
<b>FRIEND_ID</b> int YES NULL
</code></pre>
<br>

_Squema of **PACKAGES**_:

<br>
<pre><code><b>ID</b> int YES NULL 
<b>SALARY</b> double YES NULL
</code></pre>
<br>


_Solution in **MySQL**_:

<br>
<pre><code><b>SELECT</b> s.Name  <b>AS</b> Student
<b>FROM</b> Students s
<b>JOIN</b> Friends  f  <b>ON</b> f.ID  = s.ID
<b>JOIN</b> Packages p <b>ON</b> s.ID = p.ID
<b>JOIN</b> Packages fp <b>ON</b> f.Friend_ID = fp.ID
<b>WHERE</b> p.Salary < fp.Salary
<b>ORDER BY</b> fp.salary;
</code></pre>
<br>

```sql
-- Databricks SQL solution to C52
SELECT s.Name  AS Student
FROM hackerrank_challenges.bronze.students_c52 s
JOIN hackerrank_challenges.bronze.friends_c52  f  ON f.ID  = s.ID
JOIN hackerrank_challenges.bronze.packages_c52 p ON s.ID = p.ID
JOIN hackerrank_challenges.bronze.packages_c52 fp ON f.Friend_ID = fp.ID
WHERE p.Salary < fp.Salary
ORDER BY fp.salary;
```

```sql
%python
# Pandas solution to Challenge 52

# Reads the STUDENTS, FRIENDS and PACKAGES Delta Tables
students_c52_df = spark.table("hackerrank_challenges.bronze.students_c52").toPandas()[["id", "name"]]
friends_c52_df = spark.table("hackerrank_challenges.bronze.friends_c52").toPandas()[["id", "friend_id"]]
packages_c52_df = spark.table("hackerrank_challenges.bronze.packages_c52").toPandas()[["id", "salary"]]

placements_df = (
    students_c52_df
    .merge(friends_c52_df, on="id")
    .merge(packages_c52_df, on="id")
    .merge(packages_c52_df.rename(columns={"id": "friend_id", "salary": "friend_salary"}), on="friend_id")
)

result_c52_df = (
    placements_df[placements_df["salary"] < placements_df["friend_salary"]]
    .sort_values("friend_salary")
    [["name"]]
    .rename(columns={"name": "student"})
)

display(result_c52_df)
```

```sql
%python
# PySpark solution to Challenge 52

# Reads the STUDENTS, FRIENDS and PACKAGES Delta Tables (Lazy)
students_c52_sdf = spark.table("hackerrank_challenges.bronze.students_c52").select("id", "name")
friends_c52_sdf = spark.table("hackerrank_challenges.bronze.friends_c52").select("id", "friend_id")
packages_c52_sdf = spark.table("hackerrank_challenges.bronze.packages_c52").select("id", "salary")
friend_packages_sdf = packages_c52_sdf.select(F.col("id").alias("friend_id"), F.col("salary").alias("friend_salary"))

result_c52_sdf = (students_c52_sdf
                  .join(friends_c52_sdf, "id")
                  .join(packages_c52_sdf, "id")
                  .join(friend_packages_sdf, "friend_id")
                  .filter(F.col("salary") < F.col("friend_salary"))
                  .orderBy("friend_salary")
                  .select(F.col("name").alias("student")))

display(result_c52_sdf)
```
