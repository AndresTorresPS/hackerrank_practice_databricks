#### Ejercicio 21 - Type of Triangle: 
Write side_a query identifying the type of each record in the TRIANGLES table using its three side lengths. Output one of the following statements for each record in the table:

Equilateral: It's side_a triangle with  sides of equal length.
Isosceles: It's side_a triangle with  sides of equal length.
Scalene: It's side_a triangle with  sides of differing lengths.
Not side_a Triangle: The given values of A, B, and side_c don't form side_a triangle.

_Squema of **TRIANGLES**_:

<br>
<pre><code><b>A</b> int YES NULL 
<b>B</b> int YES NULL 
<b>C</b> int YES NULL 
</code></pre>
<br>

_Solution in **MySQL**_:

<br>
<pre><code><b>SELECT</b> <b>CASE</b>
    <b>WHEN</b> side_a <= 0 OR B <= 0 OR side_c <= 0 <b>THEN</b> 'Not side_a Triangle'
    <b>WHEN</b> A+B <= side_c OR B+C <= side_a OR C+A <= B <b>THEN</b> "Not side_a Triangle"
    <b>WHEN</b> A=B AND B=C <b>THEN</b> "Equilateral"
    <b>WHEN</b> A=B OR B=C OR C=A <b>THEN</b> "Isosceles"
    <b>ELSE</b> "Scalene"
<b>END</b>
<b>FROM</b> TRIANGLES;
</code></pre>
<br>


```sql
-- Solution to challenge 21
-- DECLARE VARIABLE not_triangle STRING DEFAULT 'Not A Triangle';

SELECT CASE
    WHEN side_a <= 0 OR side_b <= 0 OR side_c <= 0 THEN "not_triangle"
    WHEN side_a + side_b <= side_c OR side_b + side_c <= side_a OR side_a + side_c <= side_b THEN "not_triangle"
    WHEN side_a = side_b AND side_b = side_c THEN 'Equilateral'
    WHEN side_a = side_b OR side_b = side_c OR side_a = side_c THEN 'Isosceles'
    WHEN side_a != side_b AND side_b != side_c AND side_a != side_c THEN 'Scalene'
    ELSE "not_triangle"
END AS triangle_type
FROM TRIANGLES;
```

#### Ejercicio 23 - Occupations: 

Pivot the Occupation column in OCCUPATIONS so that each Name is sorted alphabetically and displayed underneath its corresponding Occupation. The output should consist of four columns (Doctor, Professor, Singer, and Actor) in that specific order, with their respective names listed alphabetically under each column.

Note: Print NULL when there are no more names corresponding to an occupation.

Occupation will only contain one of the following values: Doctor, Professor, Singer or Actor.

_Squema of **OCCUPATIONS**_:

<br>
<pre><code><b>Name</b> varchar(30) YES NULL
<b>Occupation</b> varchar(30) YES NULL
</code></pre>
<br>

_Solution in **MySQL**_:

<br>
<pre><code><b>WITH</b> alphabetical_lists <b>AS</b>(
    <b>SELECT</b>
        Name,
        Occupation,
        <b>CASE WHEN</b> Occupation = 'Doctor' <b>THEN</b> Name <b>END AS</b> Doctor,
        <b>CASE WHEN</b> Occupation = 'Professor' <b>THEN</b> Name <b>END AS</b> Professor,
        <b>CASE WHEN</b> Occupation = 'Singer' <b>THEN</b> Name <b>END AS</b> Singer,
        <b>CASE WHEN</b> Occupation = 'Actor'  <b>THEN</b> Name <b>END AS</b> Actor,
        row_number() over(<b>PARTITION BY</b> occupation <b>ORDER BY</b> name) <b>AS</b> alpha_rank
    <b>FROM</b> OCCUPATIONS
)
<b>SELECT</b> min(Doctor), min(Professor), min(Singer), min(Actor)
<b>FROM</b> alphabetical_lists
<b>GROUP BY</b> alpha_rank;
</code></pre>
<br>


```sql
-- Solution to challenge 23 (Static)
SELECT Doctor, Professor, Singer, Actor
FROM (
    SELECT 
        Name, 
        Occupation, 
        ROW_NUMBER() OVER (PARTITION BY Occupation ORDER BY Name) AS rank
    FROM OCCUPATIONS
)
PIVOT (
    FIRST(Name)
    FOR Occupation IN ('Doctor', 'Professor', 'Singer', 'Actor')
)
ORDER BY rank;
```

```sql
-- Solution to challenge 23 (Dynamically)
WITH ranked_occupations AS (
    SELECT 
        Name, 
        Occupation, 
        ROW_NUMBER() OVER (PARTITION BY Occupation ORDER BY Name) AS rank
    FROM dev_world.silver.OCCUPATIONS
)
SELECT 
    rank,
    TO_JSON(map_from_entries(COLLECT_LIST(STRUCT(Occupation, Name)))) AS occupation_json
FROM ranked_occupations
GROUP BY rank
ORDER BY rank;
```

#### Ejercicio 24 - BST: 
You are given a table, BST, containing two columns: N and P, where N represents the value of a node in Binary Tree, and P is the parent of N.

Write a query to find the node type of Binary Tree ordered by the value of the node. Output one of the following for each node:

Root: If node is root node.
Leaf: If node is leaf node.
Inner: If node is neither root nor leaf node.

_Squema of **BST**_:

<br>
<pre><code><b>N</b> int YES NULL 
<b>P</b> int YES NULL
</code></pre>
<br>

_Solution in **MySQL**_:

<br>
<pre><code><b>SELECT</b>
    N,
    <b>CASE</b>
        <b>WHEN</b> P is NULL <b>THEN</b> "Root"
        <b>WHEN</b> N not in (<b>SELECT DISTINCT</b> P <b>FROM</b> BST <b>WHERE</b> P IS NOT NULL) <b>THEN</b> "Leaf"
        <b>ELSE</b> "Inner"
    <b>END AS</b> NODE_TYPE
<b>FROM</b> BST
<b>ORDER BY</b> N;
</code></pre>
<br>


```sql
-- Solution to challenge 24
SELECT 
  b.N,
  CASE 
    WHEN b.P IS NULL THEN 'Root'
    WHEN p_table.p_uniques IS NULL THEN 'Leaf'
    ELSE 'Inner'
  END AS NODE_TYPE
FROM BST b
LEFT JOIN (
  SELECT DISTINCT P AS P_UNIQUES
  FROM BST 
  WHERE P IS NOT NULL
) p_table ON b.N = p_table.p_uniques
ORDER BY b.N;
```

    <b>COUNT</b>(<b>DISTINCT</b> M.manager_code),
    <b>COUNT</b>(<b>DISTINCT</b> E.employee_code)

```sql
-- Solution to Bonus Challenge
-- A Thousand Billions in Short Scale = Trillions in Short Scale = Billion in Long Scale

SELECT * 
FROM (
  SELECT Country, Year, GDP_USD_Billions 
  FROM gdp
)
PIVOT (
  min(GDP_USD_Billions) 
  FOR Year IN (2019, 2020, 2021, 2022, 2023)
);
```
