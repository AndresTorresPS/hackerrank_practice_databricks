## Ejercicios SQL de Hackerrank desde Databricks

#### Ejercicio 1: 
Query all columns for all American cities in the CITY table with populations larger than 100000. The CountryCode for America is USA.

_Squema of **CITY**_:

<br>
<pre><code><b>id</b> int YES NULL 
<b>name</b> varchar(17) YES NULL 
<b>countrycode</b> varchar(3) YES NULL 
<b>district</b> varchar(20) YES NULL 
<b>population</b> int YES NULL
</code></pre>
<br>

_Solution in **MySQL**_:

<br>
<pre><code><b>SELECT</b> * 
<b>FROM</b> CITY
<b>WHERE</b> COUNTRYCODE = "USA"
<b>AND</b> POPULATION > 100000;
</code></pre>
<br>

```sql
-- Now we try to do the same with Databriks syntax
-- In Databricks SQL (and standard ANSI SQL), string literals must use single quotes ('USA').
-- Spark SQL is case-insensitive by default for table and column identifiers (CITY and city refer to the same object). Case sensitivity can be toggled via spark.sql.caseSensitive

SELECT *
FROM dev_world.bronze.city
WHERE countrycode = 'USA'
AND population > 100000
```

```sql
-- Testing regexp_like (Case sensitive by default)
SELECT name
FROM dev_world.silver.city
WHERE regexp_like(name, '(?i)san');
```

```sql
-- Testing regexp_like (Case sensitive by default)
SELECT name
FROM dev_world.silver.city
WHERE regexp_like(name, '(?i)ro+');
```

```sql
-- Testing regexp_like (Case sensitive by default)
-- For consecutive letters
SELECT name
FROM dev_world.silver.city
WHERE regexp_like(name, '(?i)[a]{2}');
```

```sql
-- Testing regexp_like (Case sensitive by default)
-- For 3 appearances of the same letter

-- SELECT name
-- FROM dev_world.silver.city
-- WHERE regexp_like(name, '(?i)a.*a.*a');

-- Parameterized CTE (Clean Alternative)
WITH params AS (
    SELECT 
        'a'  AS target_char,
        '.*' AS separator,
        3    AS occurrences
)
SELECT c.name
FROM dev_world.silver.city c
CROSS JOIN params p
WHERE regexp_like(
    c.name, 
    concat(
        '(?i)', 
        repeat(concat(p.target_char, p.separator), p.occurrences - 1), 
        p.target_char
    )
);
```

```sql
SELECT 
        'a'  AS target_char,
        '.*' AS separator,
        3    AS occurrences
UNION ALL
SELECT 
        'b'  AS target_char,
        '.*' AS separator,
        2    AS occurrences
```

```sql
-- Starts and ends with vowels
WITH params AS (
    SELECT '[aeiou]' AS vowels
)
SELECT DISTINCT c.name
FROM dev_world.silver.city c
CROSS JOIN params p
WHERE c.name RLIKE CONCAT('(?i)^', p.vowels, '.*', p.vowels, '$');
```
