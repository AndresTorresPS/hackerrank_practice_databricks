#### Ejercicio 1: 
Query all columns for all American cities in the CITY table with populations larger than 100000. The CountryCode for America is USA.

_Squema of **CITY**_:

<br>
<pre><code><b>id</b> int YES NULL 
<b>name</b> varchar(17) YES NULL 
<b>countrycode</b> varchar(3) YES NULL 
<b>district</b> varchar(20) YES NULL 
<b>population</b> int YES NULL
</code></pre>
<br>

_Solution in **MySQL**_:

<br>
<pre><code><b>SELECT</b> * 
<b>FROM</b> CITY
<b>WHERE</b> COUNTRYCODE = "USA"
<b>AND</b> POPULATION > 100000;
</code></pre>
<br>

```sql
-- Now we try to do the same with Databriks syntax
-- In Databricks SQL (and standard ANSI SQL), string literals must use single quotes ('USA').
-- Spark SQL is case-insensitive by default for table and column identifiers (CITY and city refer to the same object). Case sensitivity can be toggled via spark.sql.caseSensitive

SELECT *
FROM dev_world.bronze.city
WHERE countrycode = 'USA'
AND population > 100000
```

#### Ejercicio 10: 
Query the two cities in STATION with the shortest and longest CITY names, as well as their respective lengths (i.e.: number of characters in the name). If there is more than one smallest or largest city, choose the one that comes first when ordered alphabetically.

_Squema of **STATION**_:

<br>
<pre><code><b>id</b> int YES NULL 
<b>CITY</b> varchar(21) YES NULL
<b>STATE</b> varchar(2) YES NULL 
<b>LAT_N</b> decimal(10,0) YES NULL
<b>LONG_W</b> decimal(10,0) YES NULL
</code></pre>
<br>

_Solution in **MySQL**_:

<br>
<pre><code>(<b>SELECT</b> CITY, LENGTH(CITY) <b>AS</b> city_len 
 <b>FROM</b> STATION 
 <b>ORDER BY</b> LENGTH(CITY), CITY 
 <b>LIMIT</b> 1)

<b>UNION ALL</b>

(<b>SELECT</b> CITY, LENGTH(CITY) <b>AS</b> city_len 
 <b>FROM</b> STATION 
 <b>ORDER BY</b> LENGTH(CITY) <b>DESC</b>, CITY 
 <b>LIMIT</b> 1);
 </code></pre>
<br>

```sql
-- Solution to challenge 10
-- It is possible to implement a suboptimal solution using CTEs

USE dev_world.bronze;

(SELECT CITY, LENGTH(CITY) AS city_len FROM STATION ORDER BY LENGTH(CITY) ASC, CITY ASC LIMIT 1)
UNION ALL
(SELECT CITY, LENGTH(CITY) AS city_len FROM STATION ORDER BY LENGTH(CITY) DESC, CITY ASC LIMIT 1);
```

#### Ejercicio 13: 
Query the list of CITY names from STATION which have vowels (i.e., a, e, i, o, and u) as both their first and last characters. Your result cannot contain duplicates.

_Squema of **STATION**_:

<br>
<pre><code><b>id</b> int YES NULL 
<b>CITY</b> varchar(21) YES NULL
<b>STATE</b> varchar(2) YES NULL 
<b>LAT_N</b> decimal(10,0) YES NULL
<b>LONG_W</b> decimal(10,0) YES NULL
</code></pre>
<br>

_Solution in **MySQL**_:

<br>
<pre><code><b>SET</b> @vowels = '[aeiou]';
<b>SELECT</b> <b>DISTINCT</b> CITY
<b>FROM</b> STATION
<b>WHERE</b> CITY <b>REGEXP</b> CONCAT('^',@vowels)
<b>AND</b> CITY <b>REGEXP</b> CONCAT(@vowels,'$');
</code></pre>
<br>

```sql
-- Solution to challenge 13

SELECT DISTINCT CITY FROM station
WHERE regexp_like(CITY, '(?i)^[aeiou].*[aeiou]$');
```

#### Ejercicio 14: 
Query the list of CITY names from STATION that do not start with vowels. Your result cannot contain duplicates.

_Squema of **STATION**_:

<br>
<pre><code><b>id</b> int YES NULL 
<b>CITY</b> varchar(21) YES NULL
<b>STATE</b> varchar(2) YES NULL 
<b>LAT_N</b> decimal(10,0) YES NULL
<b>LONG_W</b> decimal(10,0) YES NULL
</code></pre>
<br>

_Solution in **MySQL**_:

<br>
<pre><code><b>WITH</b> params <b>as</b> (
    <b>select</b>
        '^[^aeiou]' <b>as</b> not_vowel_start
)
<b>SELECT</b> <b>DISTINCT</b> s.city
<b>FROM</b> station s
<b>CROSS JOIN</b> params p
<b>WHERE</b> REGEXP_LIKE(s.city, p.not_vowel_start);
</code></pre>
<br>


```sql
-- Solution to challenge 14 
WITH params AS (
    SELECT '^[^aeiou]' as not_vowel_start
)
SELECT DISTINCT s.city
FROM station s
CROSS JOIN params p
WHERE s.city RLIKE CONCAT('(?i)^', p.not_vowel_start);
```
