#### Ejercicio 1: 
Query all columns for all American cities in the CITY table with populations larger than 100000. The CountryCode for America is USA.

_Squema of **CITY**_:

<br>
<pre><code><b>id</b> int YES NULL 
<b>name</b> varchar(17) YES NULL 
<b>countrycode</b> varchar(3) YES NULL 
<b>district</b> varchar(20) YES NULL 
<b>population</b> int YES NULL
</code></pre>
<br>

_Solution in **MySQL**_:

<br>
<pre><code><b>SELECT</b> * 
<b>FROM</b> CITY
<b>WHERE</b> COUNTRYCODE = "USA"
<b>AND</b> POPULATION > 100000;
</code></pre>
<br>

```sql
-- Catalog and shema is configured
USE dev_world.bronze;
```

```sql
-- Now we try to do the same with Databriks syntax
-- In Databricks SQL (and standard ANSI SQL), string literals must use single quotes ('USA').
-- Spark SQL is case-insensitive by default for table and column identifiers (CITY and city refer to the same object). Case sensitivity can be toggled via spark.sql.caseSensitive

SELECT *
FROM city
WHERE countrycode = 'USA'
AND population > 100000
```

#### Ejercicio 10: 
Query the two cities in STATION with the shortest and longest CITY names, as well as their respective lengths (i.e.: number of characters in the name). If there is more than one smallest or largest city, choose the one that comes first when ordered alphabetically.

_Squema of **STATION**_:

<br>
<pre><code><b>id</b> int YES NULL 
<b>CITY</b> varchar(21) YES NULL
<b>STATE</b> varchar(2) YES NULL 
<b>LAT_N</b> decimal(10,0) YES NULL
<b>LONG_W</b> decimal(10,0) YES NULL
</code></pre>
<br>

_Solution in **MySQL**_:

<br>
<pre><code>(<b>SELECT</b> CITY, LENGTH(CITY) <b>AS</b> city_len 
 <b>FROM</b> STATION 
 <b>ORDER BY</b> LENGTH(CITY), CITY 
 <b>LIMIT</b> 1)

<b>UNION ALL</b>

(<b>SELECT</b> CITY, LENGTH(CITY) <b>AS</b> city_len 
 <b>FROM</b> STATION 
 <b>ORDER BY</b> LENGTH(CITY) <b>DESC</b>, CITY 
 <b>LIMIT</b> 1);
 </code></pre>
<br>

```sql
-- Solution to challenge 10
-- It is possible to implement a suboptimal solution using CTEs
(SELECT CITY, LENGTH(CITY) AS city_len FROM STATION ORDER BY LENGTH(CITY) ASC, CITY ASC LIMIT 1)
UNION ALL
(SELECT CITY, LENGTH(CITY) AS city_len FROM STATION ORDER BY LENGTH(CITY) DESC, CITY ASC LIMIT 1);
```

<pre><code><b>WITH</b> params <b>AS</b> (
    <b>SELECT</b>

#### Ejercicio 21 - Type of Triangle: 
Write a query identifying the type of each record in the TRIANGLES table using its three side lengths. Output one of the following statements for each record in the table:

Equilateral: It's a triangle with  sides of equal length.
Isosceles: It's a triangle with  sides of equal length.
Scalene: It's a triangle with  sides of differing lengths.
Not A Triangle: The given values of A, B, and C don't form a triangle.

_Squema of **TRIANGLES**_:

<br>
<pre><code><b>A</b> int YES NULL 
<b>B</b> int YES NULL 
<b>C</b> int YES NULL 
</code></pre>
<br>

_Solution in **MySQL**_:

<br>
<pre><code><b>SELECT</b> <b>CASE</b>
    <b>WHEN</b> A <= 0 OR B <= 0 OR C <= 0 <b>THEN</b> 'Not A Triangle'
    <b>WHEN</b> A+B <= C OR B+C <= A OR C+A <= B <b>THEN</b> "Not A Triangle"
    <b>WHEN</b> A=B AND B=C <b>THEN</b> "Equilateral"
    <b>WHEN</b> A=B OR B=C OR C=A <b>THEN</b> "Isosceles"
    <b>ELSE</b> "Scalene"
<b>END</b>
<b>FROM</b> TRIANGLES;
</code></pre>
<br>

#### Ejercicio 25 - New Companies: 
Amber's conglomerate corporation just acquired some new companies. Each of the companies follows this hierarchy: 

<br>
<pre><code> Founder => Lead Manager => Senior Manager => Manager => Employee
</code></pre>
<br>
Given the table schemas below, write a query to print the company_code, founder name, total number of lead managers, total number of senior managers, total number of managers, and total number of employees. Order your output by ascending company_code.

**Note:**

1. The tables may contain duplicate records.
2. The company_code is string, so the sorting should not be numeric. For example, if the company_codes are C_1, C_2, and C_10, then the ascending company_codes will be C_1, C_10, and C_2.

**Input Format**

The following tables contain company data:

Company: The company_code is the code of the company and founder is the founder of the company. 

Lead_Manager: The lead_manager_code is the code of the lead manager, and the company_code is the code of the working company. 

Senior_Manager: The senior_manager_code is the code of the senior manager, the lead_manager_code is the code of its lead manager, and the company_code is the code of the working company. 

Manager: The manager_code is the code of the manager, the senior_manager_code is the code of its senior manager, the lead_manager_code is the code of its lead manager, and the company_code is the code of the working company. 

Employee: The employee_code is the code of the employee, the manager_code is the code of its manager, the senior_manager_code is the code of its senior manager, the lead_manager_code is the code of its lead manager, and the company_code is the code of the working company. 

_Squema of **Company**_:

<br>
<pre><code><b>company_code</b> varchar(5) YES NULL 
<b>founder</b> varchar(5) YES NULL 
</code></pre>
<br>

_Squema of **Lead_Manager**_:

<br>
<pre><code><b>employee_id</b> varchar(5) YES NULL  
<b>company_code</b> varchar(5) YES NULL 
</code></pre>
<br>

_Squema of **Senior_Manager**_:

<br>
<pre><code><b>senior_manager_code</b> varchar(5) YES NULL 
<b>lead_manager_code</b> varchar(5) YES NULL 
<b>company_code</b> varchar(5) YES NULL 
</code></pre>
<br>

_Squema of **Manager**_:

<br>
<pre><code><b>manager_code</b> varchar(5) YES NULL 
<b>senior_manager_code</b> varchar(5) YES NULL 
<b>lead_manager_code</b> varchar(5) YES NULL 
<b>company_code</b> varchar(5) YES NULL 
</code></pre>
<br>

_Squema of **Employee**_:

<br>
<pre><code><b>employee_code</b> varchar(5) YES NULL 
<b>manager_code</b> varchar(5) YES NULL 
<b>senior_manager_code</b> varchar(5) YES NULL 
<b>lead_manager_code</b> varchar(5) YES NULL 
<b>company_code</b> varchar(5) YES NULL
</code></pre>
<br>

_Solution in **MySQL**_:

<br>
<pre><code><b>SELECT</b> c.company_code, c.founder,
       COUNT(DISTINCT l.lead_manager_code), 
       COUNT(DISTINCT s.senior_manager_code), 
       COUNT(DISTINCT m.manager_code),
